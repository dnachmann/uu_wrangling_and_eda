```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv('ForeignGifts_edu.csv')
```

```python
df.value_counts('Gift Type')
```

The three different classes of gift type are contract, monetary gift, and real estate. A contract is a binding agreement between the donor and the university with specific instructions from the donor as to how the gift can be used. A monetary gift is a donation which can be used by the university for any purpose. A real estate gift is a land donation with the intention of having the university build on the land.

```python
df_sorted_country = df.groupby(df['Country of Giftor'])
total_by_country = df_sorted_country['Foreign Gift Amount'].sum()
total_by_country.sort_values(ascending = False)
```

```python
total_donations_country = df['Country of Giftor'].value_counts()
print(total_donations_country)
```

```python
country_avg = total_by_country / total_donations_country
country_avg.sort_values(ascending=False)
```

```python
df_sorted_uni = df.groupby(df['Institution Name'])
total_by_uni = df_sorted_uni['Foreign Gift Amount'].sum()
total_by_uni.sort_values(ascending = False)
```

```python
total_donations_uni = df['Institution Name'].value_counts()
print(total_donations_uni)
```

```python
uni_avg = total_by_uni / total_donations_uni
print(uni_avg.median())
print(total_by_uni.median())
```

```python
print(df_sorted_uni['Foreign Gift Amount'].mean().median())
print('\n')
print(df_sorted_uni['Foreign Gift Amount'].median().median())
```

```python
country_to_university = df.groupby(['Country of Giftor', 'Institution Name'])
country_to_university['Foreign Gift Amount'].sum().sort_values(ascending = False)
```
