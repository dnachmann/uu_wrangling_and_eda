Compared to similar universities, the University of Virginia has fewer total donations than average and a lower donation amount for both the mean and median donation. Additionally, it is last among these universities in largest donation. To fix this lack of foreign gifting, I propose expanding the donor base in locations UVA already receives donations from. Currently, UVA's top giftors by number of donations are from England, Sweden, Israel, and Bangladesh. Of those four, Bangladesh, Sweden, and England are the three largest donators by monetary value. The University should not make aiming for more large donations from Tanzania or Switzerland its primary goal, as those donations are infrequent and volatile compared to the smaller recurrent donations from Bangladesh, Sweden, and England. By expanding its donor frequency in these countries, the slow and steady gains will outpace the infrequent large donations offered by others. Average donations from Bangladesh, Sweden, and England total $890,000. While less than averages from countries with large, infrequent donations,these countries show more potential for growth. Assuming UVA runs an aggressive donation campaign where these three countries contribute 3 times for each contribution from Tanzania and Switzerland and all countries contribute 1.5x their donation, the total from smaller donations would overtake the larger ones in 3 donations as shown by the math below.
